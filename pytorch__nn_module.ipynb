## What is nn.Sequential?
nn.Sequential is a container in PyTorch that lets you stack multiple layers/modules in order, so the output of one becomes the input of the next — automatically, without you writing a forward method.



```python
import torch
import torch.nn as nn

class Model(nn.Module):
  def __init__(self,num_features):
    super().__init__()
    self.network = nn.Sequential(
    nn.Linear(num_features , 3),
    nn.ReLU(),
    nn.Linear(3,1),
    nn.Sigmoid(),
    )

  def forward(self,features):
    out = self.network(features)
    return out
```

```python
features = torch.rand(10,5)

model = Model(features.shape[1])

model(features)
```
